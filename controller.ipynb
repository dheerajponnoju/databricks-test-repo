```python
try:
    result = json.loads(dbutils.notebook.run("test1", 60))
    if result["status"] == "SUCCESS":
        status = result["status"]
        message = f"""Child Notebook executed successfully. 
            It's result: {result["message"]}
        """
        dbutils.notebook.exit({"status": status, "message": message})
    else:
        status = result["status"]
        message = f"""Child Notebook executed but failed. 
        It's result: {result["message"]}
        """
        dbutils.notebook.exit({"status": status, "message": message})
except Exception as e:
    status = "ERROR"
    message = f"""Child Notebook execution failed. 
        Exception: {e}
        """
    dbutils.notebook.exit({"status": status, "message": message})
    
```
